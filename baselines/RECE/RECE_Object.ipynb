# Perform Generations

```python
import torch
from diffusers import StableDiffusionPipeline
from pathlib import Path
import os

# --- 1. Configuration ---
# The main folder containing your subfolders with the models
main_models_folder = Path("/content/drive/My Drive/thesis/rece/outputs/")

# The top-level directory where all generated images will be saved
output_images_folder = Path("/content/drive/My Drive/thesis/rece/generated_images_oxford_flowers/")

# The base model used for generation
base_model_id = "CompVis/stable-diffusion-v1-4"

# Set the device (GPU is highly recommended)
device = "cuda" if torch.cuda.is_available() else "cpu"

# --- 2. Prompt and seed setup (you can change these) ---
prompt = "A portrait of Elon Musk and Amanda Seyfried"
seed = 0
species = "Elon Musk"
concept_type = "erased"
index = 0  # used in filename

# --- 3. Setup ---
output_images_folder.mkdir(parents=True, exist_ok=True)

# --- 4. Load Base Pipeline ---
print(f"Loading base pipeline: {base_model_id}...")
try:
    pipeline = StableDiffusionPipeline.from_pretrained(base_model_id, torch_dtype=torch.float16, safety_checker=None)
    pipeline.to(device)
    print("✅ Base pipeline loaded successfully.")
except Exception as e:
    print(f"❌ Failed to load base pipeline. Error: {e}")
    exit()

# --- 5. Find and Process Each Model ---
print(f"Searching for model files in: {main_models_folder}")
all_pt_files = list(main_models_folder.rglob('*.pt'))
model_files_to_process = [
    f for f in all_pt_files
    if "ep_0_tar_" in f.name or f.name == "final_unet_multi_concept.pt"
]

if not model_files_to_process:
    print("No model files found to process. Check the 'main_models_folder' path.")
else:
    print(f"Found {len(model_files_to_process)} models to process.")

# --- 6. Main Generation Loop ---
for unet_weights_path in model_files_to_process:
    print(f"\n{'='*20}\nProcessing Model: {unet_weights_path.name}\n{'='*20}")
    try:
        # A. Determine folder name from model path
        relative_path = unet_weights_path.relative_to(main_models_folder)
        top_folder_name = relative_path.parts[0]

        # B. Create output folder for this model
        model_specific_output_folder = output_images_folder / top_folder_name
        model_specific_output_folder.mkdir(parents=True, exist_ok=True)

        # C. Load the custom UNet weights
        print("Loading custom UNet weights...")
        unet_state_dict = torch.load(unet_weights_path, map_location=device)
        pipeline.unet.load_state_dict(unet_state_dict)
        print("✅ Weights loaded.")

        # D. Generate image
        print(f"Generating image for prompt: {prompt}")
        safe_species = species.replace(" ", "_").replace("/", "_").replace("-", "_")
        output_filename = f"rece_{safe_species}_{index:04d}_seed{seed}_{concept_type}_im0.png"
        output_image_path = model_specific_output_folder / output_filename

        if output_image_path.exists():
            print(f"Skipping {output_filename} as it already exists.")
        else:
            generator = torch.Generator(device=device).manual_seed(seed)
            with torch.autocast(device):
                image = pipeline(
                    prompt,
                    num_inference_steps=50,
                    guidance_scale=7.5,
                    generator=generator
                ).images[0]
            image.save(output_image_path)
            print(f"✅ Saved: {output_image_path}")

    except Exception as e:
        print(f"❌ An error occurred while processing {unet_weights_path.name}. Error: {e}")

print("\nAll tasks finished.")
```

```python
import torch
from diffusers import StableDiffusionPipeline
from pathlib import Path
import pandas as pd
import os

# --- 1. Configuration ---
# The main folder containing your subfolders with the models
main_models_folder = Path("/content/drive/My Drive/thesis/rece/outputs/")

# The folder containing the CSV configuration files
configs_folder = Path("/content/drive/My Drive/thesis/fgst-de/oxford-flowers/erasure_configs/")

# The top-level directory where all generated images will be saved
output_images_folder = Path("/content/drive/My Drive/thesis/rece/generated_images_oxford_flowers/")

# The base model used for training
base_model_id = "CompVis/stable-diffusion-v1-4"

# Set the device (GPU is highly recommended)
device = "cuda" if torch.cuda.is_available() else "cpu"

# --- 2. Setup ---
# Create the main output directory if it doesn't exist
output_images_folder.mkdir(parents=True, exist_ok=True)

# --- 3. Load Base Pipeline (ONCE for efficiency) ---
print(f"Loading base pipeline: {base_model_id}...")
try:
    pipeline = StableDiffusionPipeline.from_pretrained(base_model_id, torch_dtype=torch.float16, safety_checker=None)
    pipeline.to(device)
    print("✅ Base pipeline loaded successfully.")
except Exception as e:
    print(f"❌ Failed to load base pipeline. Please check model ID and connection. Error: {e}")
    exit()

# --- 4. Find and Process Each Model ---
print(f"Searching for model files in: {main_models_folder}")
all_pt_files = list(main_models_folder.rglob('*.pt'))
model_files_to_process = [
    f for f in all_pt_files
    if "ep_0_tar_" in f.name or f.name == "final_unet_multi_concept.pt"
]

if not model_files_to_process:
    print("No model files found to process. Check the 'main_models_folder' path.")
else:
    print(f"Found {len(model_files_to_process)} models to process.")

# --- 5. Main Generation Loop ---
for unet_weights_path in model_files_to_process:
    print(f"\n{'='*20}\nProcessing Model: {unet_weights_path.name}\n{'='*20}")

    try:
        # --- A. Determine the corresponding CSV config file ---
        relative_path = unet_weights_path.relative_to(main_models_folder)
        top_folder_name = relative_path.parts[0]
        config_csv_path = configs_folder / f"{top_folder_name}.csv"

        if not config_csv_path.exists():
            print(f"⚠️  Warning: Config file not found, skipping model.")
            print(f"   - Searched for: {config_csv_path}")
            continue

        print(f"Found matching config file: {config_csv_path.name}")
        df_prompts = pd.read_csv(config_csv_path)

        # --- B. Create the specific output subfolder for this model ---
        model_specific_output_folder = output_images_folder / top_folder_name # <-- CHANGED
        model_specific_output_folder.mkdir(parents=True, exist_ok=True)      # <-- ADDED

        # --- C. Load the custom UNet weights for this model ---
        print("Loading custom UNet weights...")
        unet_state_dict = torch.load(unet_weights_path, map_location=device)
        pipeline.unet.load_state_dict(unet_state_dict)
        print("✅ Weights loaded.")

        # --- D. Iterate through prompts in the CSV and generate images ---
        print(f"Generating {len(df_prompts)} images for this model...")
        for index, row in df_prompts.iterrows():
            prompt = row['prompt']
            seed = int(row['seed'])
            species = row['species']
            concept_type = row['type']

            # Sanitize species name for the filename
            safe_species = species.replace(" ", "_").replace("/", "_").replace("-", "_")

            # Create the specific output filename
            output_filename = f"rece_{safe_species}_{index:04d}_seed{seed}_{concept_type}_im0.png"
            output_image_path = model_specific_output_folder / output_filename

            if output_image_path.exists():
                print(f"Skipping {output_filename} as it already exists.")
                continue

            # Set the generator with the specific seed
            generator = torch.Generator(device=device).manual_seed(seed)

            # Generate the image
            with torch.autocast(device):
                image = pipeline(
                    prompt,
                    num_inference_steps=50,
                    guidance_scale=7.5,
                    generator=generator
                ).images[0]

            # Save the image
            image.save(output_image_path)

        print(f"✅ Finished generating images for {unet_weights_path.name}")

    except Exception as e:
        print(f"❌ An error occurred while processing {unet_weights_path.name}. Error: {e}")

print("\nAll tasks finished.")
```
